Nous avons donné à Python des nombres à comparer. Comme vous pouvez le voir, non seulement Python peut comparer des nombres, mais il peut aussi comparer les résultats des méthodes. Sympa, non?

Vous vous demandez pourquoi nous utilisons deux signes égal `==` pour comparer si les nombres sont égaux ? Nous utilisons un seul `=` pour attribuer des valeurs aux variables. * `x <= y` signifie : x est inférieur ou égal à y

* or – si vous utilisez l'opérateur `or`, seule l'une des comparaisons doit être vraie pour que l'ensemble de la commande soit vrai.

## Que se passe-t-il si une condition n'est pas True?

Dans les exemples précédents, le code était exécuté uniquement lorsque les conditions étaient True. Mais Python a aussi les instructions `elif` et `else`:

```python
if 5 > 2:
    print('5 is indeed greater than 2')
else:
    print('5 is not greater than 2')
```

Si 2 était un nombre supérieur à 5, la deuxième commande serait exécutée. Essayez-la ci-dessous!

Voyons comment fonctionne `elif`:

```python
name = 'Sonja'
if name == 'Ola':
    print('Hey Ola!')
elif name == 'Sonja':
    print('Hey Sonja!')
else:
    print('Hey anonymous!')
```

Voyez ce qui s'est passé? `elif` vous permet d'ajouter des conditions supplémentaires qui s'exécutent si les conditions précédentes échouent.

Vous pouvez ajouter autant d'instructions `elif` que vous le souhaitez après votre instruction if initiale. Par exemple:

```python
volume = 57
if volume < 20:
    print("It's kinda quiet.")
elif 20 <= volume < 40:
    print("It's nice for background music")
elif 40 <= volume < 60:
    print("Perfect, I can hear all the details")
elif 60 <= volume < 80:
    print("Nice for parties")
elif 80 <= volume < 100:
    print("A bit loud!")
else:
    print("My ears are hurting! :(")
```

Python parcourt chaque test en séquence jusqu'à ce qu'une des conditions soit vraie.

# Fonctions

Vous vous souvenez des fonctions comme `len()` que vous pouvez exécuter en Python ? Eh bien, bonne nouvelle – vous allez apprendre à écrire vos propres fonctions maintenant!

Une fonction est une séquence d'instructions que Python doit exécuter. Chaque fonction en Python commence par le mot-clé `def`, lui donne un nom, et peut avoir certains paramètres. Essayons:

```python
def hi():
    print('Hi there!')
    print('How are you?')

hi()
```

D'accord, notre première fonction est prête!

Vous vous demandez peut-être pourquoi nous avons écrit le nom de la fonction en bas du fichier. C'est parce que Python lit le fichier et l'exécute de haut en bas. Donc, pour utiliser notre fonction, nous devons l'écrire en haut du fichier avant de l'appeler plus bas.





>Remarque : si votre fonction ne fonctionne pas, ne paniquez pas ! Le message d'erreur vous aidera à comprendre pourquoi :
* `NameError` : Cela signifie probablement que vous avez fait une erreur de frappe, alors vérifiez que vous avez utilisé le même nom lors de la création de la fonction avec `def hi()` et lors de son appel avec `hi()`.
* `IndentationError` : Vérifiez que toutes les lignes de code à l'intérieur de la fonction ont le même niveau d'indentation au début de la ligne : Python veut que tout le code à l'intérieur de la fonction soit bien aligné.
*Pas de sortie du tout : Assurez-vous que le dernier `hi()` n'est pas indenté – si c'est le cas, cette ligne deviendra également partie de la fonction, et elle ne sera jamais exécutée.

Construisons notre première fonction avec des paramètres. Nous allons modifier l'exemple précédent – une fonction qui dit "hi" à la personne qui l'exécute – en donnant à notre fonction un paramètre appelé `name`:


```python
def hi(name):
    if name == 'Ola':
        print('Hi Ola!')
    elif name == 'Sonja':
        print('Hi Sonja!')
    else:
        print('Hi anonymous!')
hi()

# Remember: The print function is indented four spaces within the if statement.
# This is because the function runs when the condition is met.
```

Oups, une erreur. Heureusement, Python nous donne un message d'erreur assez utile. Il nous indique que la fonction `hi()` (celle que nous avons définie) a un argument requis (appelé `name`) et que nous avons oublié de le passer lors de l'appel de la fonction. Corrigeons cela:

```python
hi("Ola")
```

Et si nous changeons le nom?

```python
hi("Sonja")
```

Que pensez-vous qu'il se passera si vous écrivez un autre nom (pas Ola ou Sonja) ? Essayez et voyez si vous avez raison.

# Boucles

Les programmeurs n'aiment pas se répéter. La programmation consiste à automatiser les choses, donc nous ne voulons pas saluer chaque personne par son nom manuellement, n'est-ce pas? C'est là que les boucles sont très pratiques.

Vous vous souvenez des listes? Faisons une liste de filles:

```python
girls = ['Rachel', 'Monica', 'Phoebe', 'Ola', 'You']
```

Nous voulons les saluer toutes par leur prénom. Nous avons la fonction hi pour cela, alors utilisons-la dans une boucle:

```python
for name in girls:
    hi(name)
    print('Next girl')
```

L'instruction `for` fonctionne de manière similaire à l'instruction `if`: le code en dessous de ces deux instructions doit être indenté de quatre espaces.


Comme vous pouvez le voir, tout ce que vous mettez dans une instruction for avec une indentation sera répété pour chaque élément de la liste girls.
Vous pouvez également utiliser for sur des nombres en utilisant la fonction range:

```python
for i in range(1, 6):
    print(i)
```

`range` est une fonction qui crée une liste de nombres consécutifs (ces nombres sont fournis par vous en tant que paramètres).

Notez que le deuxième de ces deux nombres n'est pas inclus dans la liste générée par Python (ce qui signifie que `range(1, 6`) compte de 1 à 5, mais n'inclut pas le nombre 6). C'est parce que "range" est semi-ouvert, ce qui signifie qu'il inclut la première valeur, mais pas la dernière.
